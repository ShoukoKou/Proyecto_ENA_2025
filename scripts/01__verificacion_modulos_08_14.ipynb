# Verificación de los módulos 08 y 14

## 1. Objetivo

Comparamos los archivos `.parquet` de los módulos 08 (Buenas Prácticas Agrícolas) y 14 (Inocuidad) de la ENA 2025 con sus archivos Excel de origen.

Revisaremos las dimensiones, estructura de columnas, tipos de datos, valores, datos ausentes y textos vacíos, sin modificar los archivos originales.

````python
import geopandas as gpd
import pandas as pd
````

````python
# Leer los archivos Parquet
df_buenas_practicas = pd.read_parquet("../data/raw/2025/08_CAP300AB.parquet")
df_inocuidad = pd.read_parquet("../data/raw/2025/14_CAP600.parquet")

# Leer los archivos Excel originales
df_excel_08 = pd.read_excel("../data/raw/excel/08_CAP300AB.xlsx")
df_excel_14 = pd.read_excel("../data/raw/excel/14_CAP600.xlsx")
````

## 2. Comparación de dimensiones

Comparamos la cantidad de filas y columnas de los archivos ```excel``` y ```parquet``` para comprobar que la conversión a ```.parquet``` conservó sus dimensiones.

````python
print("Módulo 08")
print("Excel:", df_excel_08.shape)
print("Parquet:", df_buenas_practicas.shape)

print("\nMódulo 14")
print("Excel:", df_excel_14.shape)
print("Parquet:", df_inocuidad.shape)
````

## 3. Comparación de nombres y orden de columnas

Verificamos que los archivos ```excel``` y ```parquet``` contienen las mismas columnas, con los mismos nombres y en el mismo orden.

````python
print("Módulo 08:", df_excel_08.columns.equals(df_buenas_practicas.columns))
print("Módulo 14:", df_excel_14.columns.equals(df_inocuidad.columns))
````

## 4. Inspección estructural con `.info()`

Revisamos el número de filas, las columnas, sus tipos y cuántos valores no nulos tiene cada una. Esta vista general permite conocer la estructura antes de comparar los tipos y revisar datos ausentes.

### 4.1 Información de Buenas Prácticas

````python
df_buenas_practicas.info(verbose=True, show_counts=True)
````

Son 39729 registros en total, además de 111 columnas. De aquellas, 33 no presentan valores nulos, mas las 78 restantes tienen al menos un valor faltante.

### 4.2 Información de Inocuidad

````python
df_inocuidad.info(verbose=True, show_counts=True)
````

Son 40020 registros en total, además de 61 columnas. De aquellas, 13 no presentan valores nulos, mas las 48 restantes tienen al menos un valor faltante.

## 5. Comparación de tipos de datos

Comparamos los tipos de datos de cada columna entre los archivos Excel y Parquet para identificar alguna posible diferencia de representación.

````python
# .dtypes muestra el tipo de cada columna
print("Módulo 08:", df_excel_08.dtypes.equals(df_buenas_practicas.dtypes))

print("Módulo 14:", df_excel_14.dtypes.equals(df_inocuidad.dtypes))
````

## 6. Comparación de valores y datos ausentes

Comprobamos si ambas tablas coinciden por completo. `DataFrame.equals()` compara también la ubicación de los valores ausentes.

````python
print("Módulo 08:", df_excel_08.equals(df_buenas_practicas))
print("Módulo 14:", df_excel_14.equals(df_inocuidad))
````

## 7. Identificación de textos vacíos y espacios

Buscamos en todas las columnas si hay celdas de texto vacías o compuestas solo por espacios. `strip()` reconoce también éstas últimas.

### 7.1 Módulo de Buenas Prácticas

````python
df_excel_void_08 = df_excel_08.map(lambda valor: isinstance(valor, str) and valor.strip()=="")

esp_column_08 = df_excel_void_08.sum()

print("Módulo 08:")
print(esp_column_08[esp_column_08 > 0])
````

### 7.2 Módulo de Inocuidad

````python
df_excel_void_14 = df_excel_14.map(lambda valor: isinstance(valor, str) and valor.strip()=="")

esp_column__14 = df_excel_void_14.sum()

print("Módulo 14:")
print(esp_column__14[esp_column__14 > 0])
````

## 8. Resumen de valores ausentes

Identificamos la cantidad total de celdas con valores nulos y todas las columnas que presentan al menos un valor ausente en cada módulo, ordenadas de mayor a menor.

### 8.1 Módulo de Buenas Prácticas

````python
nulos_08 = df_buenas_practicas.isna().sum()

print("Módulo 08")
print("Total de celdas nulas:", nulos_08.sum())
print("Columnas con valores nulos:", (nulos_08 > 0).sum())

print("\nValores nulos por columna:")
print(nulos_08[nulos_08 > 0].sort_values(ascending=False).to_string())
````

### 8.2 Módulo de Inocuidad

````python
nulos_14 = df_inocuidad.isna().sum()

print("Módulo 14")
print("Total de celdas nulas:", nulos_14.sum())
print("Columnas con valores nulos:", (nulos_14 > 0).sum())

print("\nValores nulos por columna:")
print(nulos_14[nulos_14 > 0].sort_values(ascending=False).to_string())
````

### 8.3 Columnas sin valores nulos

Identificamos las columnas que tienen valores no nulos en todos sus registros para ambos módulos.

````python
for nombre, df in [("Módulo 08", df_buenas_practicas),
                   ("Módulo 14", df_inocuidad)]:
    columnas_completas = df.columns[df.notna().all()].tolist()

    print(f"\n{nombre}: {len(columnas_completas)} columnas sin nulos:")
    print(columnas_completas)
````

## 9. Conclusión

La comparación de los archivos Excel y Parquet de los módulos 08 y 14 permitió verificar lo siguiente:

- Las dimensiones de los archivos de Excel y Parquet coinciden en sus respectivos módulos.
- Los nombres y el orden de las columnas también coinciden.
- No se identificaron diferencias en los tipos de datos leídos por pandas.
- La comparación completa por `DataFrame.equals()` retornó `True` para ambos módulos, incluidos los datos ausentes en las mismas posiciones.
- Las columnas `OMICAP300` y `OMICAP600` contienen cadenas vacías u espacios en blanco en todos sus registros.

La verificación no afirma que todos los datos se encuentren limpios. La interpretación de variables, códigos de respuesta, valores ausentes y posibles necesidades de limpieza se realizará durante la exploración de cada módulo. Adicionalmente, no se modificaron los archivos originales.
